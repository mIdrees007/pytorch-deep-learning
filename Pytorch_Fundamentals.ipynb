### Matric MULTIPLICATION
Toe main ways of performing multiplication in neural network and deep learning.
1. Element wise multiplication

*   List item
*   List item


2. Matrix multiplication



There are two rules that performing matirx multiplication needs to satisfy;
1. The **inner dimensions** must match


*   (3, 2) @ (3, 2) won't work
*   (2, 3) @ (3, 2) will work


*   (3, 2) @ (2, 3) will work
2. The resulting matrix has the shape of outer dimensions


*  (2, 3) @ (3, 2) -> (2, 2)










```python
tensor = torch.tensor([1, 2, 3])
```

```python
# Element wise multiplication
print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# Matrix Multiplicatin
torch.matmul(tensor, tensor)
```

```python
%%time
value  = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]
print(value)
```

```python
%%time
torch.matmul(tensor, tensor)
```

### One of the most common errors in deep learning

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

#torch.mm(tensor_A, tensor_B) # torch.mm is the same as torch.matmul (it's an alias for matmul)
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tensor issues , we can maipulate the shape of one of our tensors using  A *transpose* switches the axes of dimensions

```python
tensor_B.T, tensor_B.T.shape
```

```python
tensor_B, tensor_B.shape
```

```python
# The mtrix multiplication operaration works when tensor B is transposed.
print(f"Orignal shapes: tensor_A {tensor_A.shape}, tensor_B ={tensor_B.shape}")
print(f"New Shapes : tensor_A: {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}")
print(f"Multiplication: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match")
print("Output: \n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```

### Reshaping , stacking, squeezing, and unsqueezing tensors


Reshaping- reshapes an input tensor to a defined shape
View-Return a view of an input tensor of =certain but keep the same memory as the orginal tensor
Stacking- combine multiple tensors on top of each other (vstack) or side by side hstack
Squeeze-> removes all 1 dimensions form a tensor
Unsqueeze-> Add a 1 dimension to a target tensor
Permute--> Return a view of the input with dimension permuted (swapped) in a certain way





```python
# create tensor
x2 = torch.arange(1., 10.)
x2, x2.shape
```

```python
# Add an extra dimension
x2_reshaped = x2.reshape(1, 9)
x2_reshaped, x2_reshaped.shape
```

```python
x2_reshaped = x2.reshape(9, 1)
x2_reshaped, x2_reshaped.shape
```

```python
#changing the view
z = x2.view(1, 9)
z, z.shape
```

# Changing z changes x (because a view of a tensor shares the same memory as the orignal)

```python
z[:, 0] = 5
z, x2
```

```python
# stack tensors on top of each other
x_stacked = torch.stack([x2, x2, x2], dim=0)
x_stacked
```

```python
# torch.squeeze() -removes all single dimensions from a target tensor
print(f"previous tesnor: {x2_reshaped}")
```

```python
print(f"previous shape: {x2_reshaped.shape}")
x2_reshaped.shape
```

```python
# Remove extra dimensions from x2_reshaped
x2_squeezed = x2_reshaped.squeeze()
print(f"\nNew tensor: {x2_squeezed}")
print(f"New shape: {x2_squeezed.shape}")
```

```python
# torch.unsqueezed() -adds a single dimension to a target tensor at a specific dim(dimension)
print(f"Previous target: {x2_squeezed}")
print(f"previous shape: {x2_squeezed.shape}")

# Add an extra dimension with unsqueeze
x2_unsqueezed = x2_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x2_unsqueezed}")
print(f"\nNew Shape : {x2_unsqueezed.shape}")
```

```python
# torch.permute --mean rearrange

x_orignal = torch.rand(size=(224, 224, 3)) # [height, width, color_channels]

# Permute the orignal tensor to rearrange the axis or dim order
x_permuted = x_orignal.permute(2, 0, 1) # shifts 0>1, 1>2, 2>0

print(f"Previous shape:{x_orignal.shape}")
print(f"New shape: {x_permuted.shape}") # [color_channels, height, width]
```

### Pytorch and Numpy
Numpy is a popular scientific Python numerical computing library
And because of this, pytorch has funcitonality to interact with it
* Data in Numpy, want in PyTorch tensor --> `torch.from_numpy(ndarray)`
* PyTorch tensor-> Numpy-> torch.Tensor.numpy()

```python
# Numpy array to tensor
import numpy as np

array  = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
array, tensor
```

```python
array.dtype
```

```python
tensor.dtype
```

```python
torch.arange(1.0, 8.0).dtype
```

```python
# Change the value of array what will this do to `tensor`
array = array + 1
array, tensor
```

```python
# Tensor to NUMPY ARRAY
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
numpy_tensor.dtype
```
